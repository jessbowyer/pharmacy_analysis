```python
csv_data = """Hour,Prescriptions,Staff,AvgWaitMinutes
9,35,3,8
10,50,3,12
11,60,3,15
12,70,3,18
13,65,3,17
14,55,3,14
15,75,3,20
16,95,4,25
17,120,4,35
18,100,4,30"""

file_name = "pharmacy_data.csv"

with open(file_name, "w") as f:
    f.write(csv_data)

print(f"Data successfully written to {file_name}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# load pharmacy data
df = pd.read_csv("pharmacy_data.csv")

# show table
print(df)

# plot prescriptions by hour
plt.plot(df["Hour"], df["Prescriptions"])
plt.title("Prescriptions by Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Prescriptions")
plt.show()

# plot wait time
plt.plot(df["Hour"], df["AvgWaitMinutes"])
plt.title("Waiting Time by Hour")
plt.xlabel("Hour")
plt.ylabel("Minutes")
plt.show()
```

# Pharmacy Busy Times Analysis

## Objective
Analyse pharmacy prescription volume and waiting times to identify peak operational periods.

## Method
Used Python (Pandas + Matplotlib) to visualise hourly prescription data.

## Results
Peak prescription load occurred between 4pm–6pm, with waiting times increasing accordingly.

## Conclusion
Data suggests increasing staffing during peak hours could reduce patient waiting times.
