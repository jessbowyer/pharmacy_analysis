```python
csv_data = """Reason,Count
Similar drug name,18
Wrong strength selected,25
Wrong patient label,12
Picking error,30
Interruptions,20
Packaging confusion,15
Unreadable prescription,10"""

file_name = "near_misses.csv"

with open(file_name, "w") as f:
    f.write(csv_data)

print(f"Data successfully written to {file_name}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# load data
df = pd.read_csv("near_misses.csv")

print(df)

# bar chart of causes
plt.bar(df["Reason"], df["Count"])
plt.xticks(rotation=60)
plt.title("Pharmacy Near Miss Causes")
plt.ylabel("Number of Incidents")
plt.tight_layout()
plt.show()
```
